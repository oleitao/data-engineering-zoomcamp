# dlt Homework - Solution

This notebook solves the exercises from `dlt_homework.md` using paginated API reads and pandas calculations.

Source: `https://us-central1-dlthub-analytics.cloudfunctions.net/data_engineering_zoomcamp_api`

## 1) Data setup and loading (commands in sequence)

```python
import requests
import pandas as pd

BASE_URL = "https://us-central1-dlthub-analytics.cloudfunctions.net/data_engineering_zoomcamp_api"
```

```python
# Command 1: paginate until an empty page is returned
rows = []
page = 1

while True:
    data = requests.get(BASE_URL, params={"page": page}, timeout=60).json()
    if not data:
        break
    rows.extend(data)
    page += 1

print(f"Pages read: {page - 1}")
print(f"Total records: {len(rows)}")
```

```python
# Command 2: create dataframe and prepare datetime columns
df = pd.DataFrame(rows)
df["Trip_Pickup_DateTime"] = pd.to_datetime(df["Trip_Pickup_DateTime"])
df["Trip_Dropoff_DateTime"] = pd.to_datetime(df["Trip_Dropoff_DateTime"])

df.shape
```

## 2) Question 1 - Dataset start date and end date

Commands used:
1. `pickup_min = df["Trip_Pickup_DateTime"].min()`
2. `dropoff_max = df["Trip_Dropoff_DateTime"].max()`
3. Extract only the date part with `.date()`

```python
pickup_min = df["Trip_Pickup_DateTime"].min()
dropoff_max = df["Trip_Dropoff_DateTime"].max()

print("Start (pickup):", pickup_min.date())
print("End (dropoff):", dropoff_max.date())
```

**Answer Q1:** `2009-06-01 to 2009-07-01`

## 3) Question 2 - Proportion of trips paid by credit card

Commands used:
1. Normalize `Payment_Type` text
2. Count rows where value is `credit`
3. Divide by total and multiply by 100

```python
payment_clean = df["Payment_Type"].astype(str).str.strip().str.lower()
credit_count = (payment_clean == "credit").sum()
credit_pct = round((credit_count / len(df)) * 100, 2)

print("Trips paid with credit:", credit_count)
print("Credit percentage (%):", credit_pct)
```

**Answer Q2:** `26.66%`

## 4) Question 3 - Total amount of money generated in tips

Commands used:
1. Convert `Tip_Amt` to numeric
2. Sum all values
3. Round to 2 decimal places

```python
tips_total = pd.to_numeric(df["Tip_Amt"], errors="coerce").sum()
tips_total = round(float(tips_total), 2)

print("Total tips ($):", tips_total)
```

**Answer Q3:** `$6,063.41`

## 5) Final answers (multiple choice)

1. `2009-06-01 to 2009-07-01`
2. `26.66%`
3. `$6,063.41`
